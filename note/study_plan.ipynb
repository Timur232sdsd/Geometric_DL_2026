```python
class DSU:
    def __init__(self, size: int):
        self.parent = [None] * (size + 1)
    def find(self, i: int):
        if self.parent[i] is not None:
            return self.find(self.parent[i])
        else:
            return i
    def union(self, i: int, j: int) -> None:
        x = self.find(i)
        y = self.find(j)
        if x != y:
            self.parent[x] = y

```

```python
if __name__ == "__main__":
    dsu = DSU(5)
    dsu.union(1, 2)
    dsu.union(2, 3)
    dsu.union(4, 5)
    
    print(dsu.find(1))
    print(dsu.find(2))
```
